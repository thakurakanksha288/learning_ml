```python
df4=df3.copy()
```

```python
df4['total_sqft']=df4['total_sqft'].apply(lambda x: x.replace(',', ''))
```

```python
df4.head()
```

```python
df5=df4.copy()

def convert_sqft_to_num(x):
    x = str(x).strip()
    tokens = x.split('-')
    if len(tokens) == 2:
        try:
            return (float(tokens[0]) + float(tokens[1])) / 2
        except ValueError:
            return None
    try:
        return float(x)
    except ValueError:
        return None

df5['total_sqft'] = df5['total_sqft'].apply(convert_sqft_to_num)

# Drop rows where total_sqft could not be converted to a number (resulted in None/NaN)
df5.dropna(subset=['total_sqft'], inplace=True)

df5['price_per_shift']=df5['price']*100000/df5['total_sqft']
df5.head()
```

```python
len(df5.location.unique())
```

```python
df5.location=df5.location.apply(lambda x: x.strip())
location_stats=df5.groupby('location')['location'].agg('count')


location_stats
```

```python
location_stats_less_than_10=len(location_stats[location_stats<=10])

location_stats_less_than_10
```

```python
len(df5.location.unique())
```

```python
location_stats_less_than_10_names = location_stats[location_stats<=10].index
df5.location = df5.location.apply(lambda x: 'other' if x in location_stats_less_than_10_names else x )

len(df5.location.unique())
```

```python
def remove_pps_outliers(df):
  df_out=pd.DataFrame()
  for key,subdf in df.groupby('location'):
    m=np.mean(subdf.price_per_shift)
    st=np.std(subdf.price_per_shift )
    reduced_df=subdf[(subdf.price_per_shift>(m-st)) & (subdf.price_per_shift<=(m+st))]
    df_out=pd.concat([df_out,reduced_df],ignore_index=True)
  return df_out
```

```python
df7=remove_pps_outliers(df6)
df7.head()
df7.shape
```
